```python
import pandas as pd
from spicy import io
import numpy as np
import os
import mat73
```

```python
data = mat73.loadmat("data\data\MAT files\Participant1\Processed_Data.mat")
```

```python
data = data["Subject1_pro"]
```

```python
import numpy as np

def unwrap(x):
    """Recursively unwrap matlab object arrays"""
    while isinstance(x, np.ndarray) and x.dtype == object:
        if x.size == 1:
            x = x.item()
        else:
            break
    return x
```

```python
data.dtype.names
```

```python
angles = unwrap(unwrap(unwrap(data["V1"])['R'])['Angles'])
angles_flat = np.array(angles).flatten()

# unwrap từng phần tử nếu cần
angles_flat = [unwrap(x) for x in angles_flat]
```

```python
import pandas as pd

rows = []

for speed_name in data.dtype.names:   # V1, V15...
    speed_block = unwrap(data[speed_name])

    for side in ['R', 'L']:
        side_block = unwrap(speed_block[side])

        # iterate trials
        for trial_idx in range(side_block.shape[1]):

            trial = unwrap(side_block[0, trial_idx])

            trial_name = unwrap(trial[0])
            table = unwrap(trial[1])

            row = {
                "speed": speed_name,
                "side": side,
                "trial": trial_name
            }

            # extract fields inside table
            for field in table.dtype.names:
                value = unwrap(table[field])
                row[field] = value

            rows.append(row)

df = pd.DataFrame(rows)
```
